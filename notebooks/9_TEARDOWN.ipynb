# Module 9: Teardown - Clean Up Lab Environment

## Purpose
Remove all objects created during the Hands-On Lab. Run this ONLY when you're finished and want to clean up your Snowflake account.

> **WARNING:** This will permanently DROP the database, roles, and all data created in this lab. This action cannot be undone.

---

> **Role Required:** `ACCOUNTADMIN`

> **What this does:** Switches to ACCOUNTADMIN role which has the privileges needed to drop databases, roles, and integrations.


```sql
-- Switch to ACCOUNTADMIN for cleanup
USE ROLE ACCOUNTADMIN;
```

---
## Step 1: Suspend Active Objects

Stop any running alerts and tasks before dropping.

> **What this does:** Suspends the DQ alert and scheduled task to stop ongoing compute before dropping objects.


```sql
-- Suspend alert and task (ignore errors if they don't exist)
ALTER ALERT IF EXISTS CORP_DWH.DQ.ALERT_DQ_FAILURES SUSPEND;
ALTER TASK IF EXISTS CORP_DWH.DQ.TASK_HOURLY_DQ_SWEEP SUSPEND;
```

---
## Step 2: Drop the Database

This removes ALL schemas, tables, views, DMFs, procedures, and data.

> **What this does:** Drops the entire CORP_DWH database, removing all schemas, tables, views, and procedures created during the lab.


```sql
DROP DATABASE IF EXISTS CORP_DWH;
DROP WAREHOUSE IF EXISTS DQ_LAB_WH;
```

---
## Step 3: Drop Roles

> **What this does:** Drops the lab roles (CORP_DQ_ADMIN and CORP_DQ_STEWARD) to clean up RBAC objects.


```sql
DROP ROLE IF EXISTS CORP_DQ_ADMIN;
DROP ROLE IF EXISTS CORP_DQ_STEWARD;
```

---
## Step 4: Drop Notification Integration

> **What this does:** Drops the notification integration used for DQ alert emails.


```sql
DROP INTEGRATION IF EXISTS CORP_DQ_ALERTS;
```

---
## Step 5: Verify Cleanup

> **What this does:** Verifies the database was dropped by checking if it still appears in SHOW DATABASES.


```sql
-- These should all return empty or error
SHOW DATABASES LIKE 'CORP_DWH';
```

> **What this does:** Verifies the roles were dropped by checking if any CORP_DQ roles still exist.


```sql
SHOW ROLES LIKE 'CORP_DQ%';
```

> **What this does:** Verifies your work so far. All checks should show [PASS].


```sql
from snowflake.snowpark.context import get_active_session
session = get_active_session()

print("=" * 50)
print("TEARDOWN VERIFICATION")
print("=" * 50)

# Check database is gone
dbs = session.sql("SHOW DATABASES LIKE 'CORP_DWH'").collect()
if len(dbs) == 0:
    print("  [PASS] Database CORP_DWH dropped successfully")
else:
    print("  [FAIL] Database CORP_DWH still exists")

# Check roles are gone
roles = session.sql("SHOW ROLES LIKE 'CORP_DQ%'").collect()
if len(roles) == 0:
    print("  [PASS] Roles CORP_DQ_ADMIN and CORP_DQ_STEWARD dropped")
else:
    print(f"  [FAIL] {len(roles)} CORP_DQ roles still exist")

print("\n  Cleanup complete. Your account is back to its original state.")
print("=" * 50)
```
